```python
from pathlib import Path
import pandas as pd

data_path = Path("../data/raw/questionnaire/CLEAN- PCOS SURVEY SPREADSHEET.csv")

df = pd.read_csv(data_path)

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("Total columns:", len(df.columns))

for i, col in enumerate(df.columns, start=1):
    print(i, repr(col))
```

```python
print("Dataset shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())

print("\nColumn data types:")
display(df.dtypes.to_frame(name="Data Type"))

print("\nDataset information:")
df.info()
```

```python
print("Missing values in each column:")
display(df.isnull().sum().to_frame(name="Missing Values"))

print("\nTotal missing values:", df.isnull().sum().sum())

print("\nExact duplicate rows:", df.duplicated().sum())

print("\nNumber of unique values in each column:")
display(df.nunique().to_frame(name="Unique Values"))
```

```python
for column in df.columns:
    print(f"\n{column}")
    print(df[column].value_counts().sort_index().to_dict())
```

```python
height_col = "Height (in Cm / Feet)"
age_col = "Age (in Years)"
duration_col = "How long does your period last ? (in Days)\r\nexample- 1,2,3,4....."

print("Rows with unusual height values:")
display(df[df[height_col] < 100][[age_col, height_col]])

print("\nRows with ages outside the stated 18–45 range:")
display(df[(df[age_col] < 18) | (df[age_col] > 45)][[age_col, height_col]])

print("\nRows with unusual period-duration values (0, 14, or 15):")
display(df[df[duration_col].isin([0, 14, 15])][[age_col, duration_col]])
```

```python
symptom_col = "Do have hair loss/hair thinning/baldness ?"

counts = df[symptom_col].value_counts().sort_index()

plt.figure(figsize=(6, 4))
counts.plot(kind="bar")

plt.title("Recorded Responses: Hair Loss / Hair Thinning")
plt.xlabel("Recorded Response Code")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)

for position, count in enumerate(counts.values):
    plt.text(position, count + 2, str(count), ha="center")

plt.tight_layout()
plt.show()
```

```python
from pathlib import Path

output_dir = Path("../outputs/questionnaire")
output_dir.mkdir(parents=True, exist_ok=True)

plt.figure(figsize=(6, 4))
counts.plot(kind="bar")

plt.title("Recorded Responses: Hair Loss / Hair Thinning")
plt.xlabel("Recorded Response Code")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)

for position, count in enumerate(counts.values):
    plt.text(position, count + 2, str(count), ha="center")

plt.tight_layout()
plt.savefig(output_dir / "hair_loss_response_frequency.png", dpi=300, bbox_inches="tight")
plt.show()

print("Graph saved to:", output_dir / "hair_loss_response_frequency.png")
```

```python
missing_counts = df.isnull().sum()

plt.figure(figsize=(12, 5))
missing_counts.plot(kind="bar")

plt.title("Missing Values by Questionnaire Variable")
plt.xlabel("Questionnaire Variables")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=90, fontsize=8)

plt.tight_layout()
plt.savefig(output_dir / "missing_values_summary.png",
            dpi=300, bbox_inches="tight")
plt.show()

print("Total missing values:", missing_counts.sum())
print("Graph saved to:", output_dir / "missing_values_summary.png")
```

```python

keywords = [
    "blood group",
    "diagnosed with PCOS",
    "After how many months",
    "periods regular",
    "period last"
]

for keyword in keywords:
    matches = [c for c in df.columns if keyword.lower() in c.lower()]

    for col in matches:
        print(f"\n--- {col.strip()} ---")
        print(df[col].value_counts(dropna=False).to_string())
```
