```python
from pathlib import Path
import pandas as pd

data_path = Path("../data/raw/questionnaire/CLEAN- PCOS SURVEY SPREADSHEET.csv")

df = pd.read_csv(data_path)

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("Dataset shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())

print("\nColumn data types:")
display(df.dtypes.to_frame(name="Data Type"))

print("\nDataset information:")
df.info()
```

```python
print("Missing values in each column:")
display(df.isnull().sum().to_frame(name="Missing Values"))

print("\nTotal missing values:", df.isnull().sum().sum())

print("\nExact duplicate rows:", df.duplicated().sum())

print("\nNumber of unique values in each column:")
display(df.nunique().to_frame(name="Unique Values"))
```

```python
for column in df.columns:
    print(f"\n{column}")
    print(df[column].value_counts().sort_index().to_dict())
```

```python
height_col = "Height (in Cm / Feet)"
age_col = "Age (in Years)"
duration_col = "How long does your period last ? (in Days)\r\nexample- 1,2,3,4....."

print("Rows with unusual height values:")
display(df[df[height_col] < 100][[age_col, height_col]])

print("\nRows with ages outside the stated 18–45 range:")
display(df[(df[age_col] < 18) | (df[age_col] > 45)][[age_col, height_col]])

print("\nRows with unusual period-duration values (0, 14, or 15):")
display(df[df[duration_col].isin([0, 14, 15])][[age_col, duration_col]])
```

```python
symptom_col = "Do have hair loss/hair thinning/baldness ?"

counts = df[symptom_col].value_counts().sort_index()

plt.figure(figsize=(6, 4))
counts.plot(kind="bar")

plt.title("Recorded Responses: Hair Loss / Hair Thinning")
plt.xlabel("Recorded Response Code")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)

for position, count in enumerate(counts.values):
    plt.text(position, count + 2, str(count), ha="center")

plt.tight_layout()
plt.show()
```

```python
from pathlib import Path

output_dir = Path("../outputs/questionnaire")
output_dir.mkdir(parents=True, exist_ok=True)

plt.figure(figsize=(6, 4))
counts.plot(kind="bar")

plt.title("Recorded Responses: Hair Loss / Hair Thinning")
plt.xlabel("Recorded Response Code")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)

for position, count in enumerate(counts.values):
    plt.text(position, count + 2, str(count), ha="center")

plt.tight_layout()
plt.savefig(output_dir / "hair_loss_response_frequency.png", dpi=300, bbox_inches="tight")
plt.show()

print("Graph saved to:", output_dir / "hair_loss_response_frequency.png")
```

```python
missing_counts = df.isnull().sum()

plt.figure(figsize=(12, 5))
missing_counts.plot(kind="bar")

plt.title("Missing Values by Questionnaire Variable")
plt.xlabel("Questionnaire Variables")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=90, fontsize=8)

plt.tight_layout()
plt.savefig(output_dir / "missing_values_summary.png",
            dpi=300, bbox_inches="tight")
plt.show()

print("Total missing values:", missing_counts.sum())
print("Graph saved to:", output_dir / "missing_values_summary.png")
```

```python
df.columns = df.columns.str.replace(r"[\r\n]+", " ", regex=True).str.strip()
```

```python
target_col = "Have you been diagnosed with PCOS/PCOD?"

target_counts = df[target_col].value_counts().sort_index()

plt.figure(figsize=(6, 4))
target_counts.plot(kind="bar")

plt.title("Recorded PCOS/PCOD Diagnosis Responses")
plt.xlabel("Recorded Response Code")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)

for position, count in enumerate(target_counts.values):
    plt.text(position, count + 2, str(count), ha="center")

plt.tight_layout()
plt.savefig(
    output_dir / "pcos_diagnosis_response_distribution.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

print("Response counts:")
print(target_counts)
print("Graph saved successfully!")
```

```python
print("QUESTIONNAIRE DATASET — EDA SUMMARY")
print("-" * 45)
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
print(f"Total missing values: {df.isnull().sum().sum()}")
print(f"Exact duplicate rows: {df.duplicated().sum()}")

print("\nAge range in dataset:")
print(f"Minimum age: {df['Age (in Years)'].min()}")
print(f"Maximum age: {df['Age (in Years)'].max()}")

height_col = "Height (in Cm / Feet)"
print("\nHeight entries below 100:", (df[height_col] < 100).sum())

duration_col = next(
    col for col in df.columns
    if col.startswith("How long does your period last")
)
print(
    "Period duration entries coded 0, 14, or 15:",
    df[duration_col].isin([0, 14, 15]).sum()
)

print("\nRecorded diagnosis codes:")
print(df["Have you been diagnosed with PCOS/PCOD?"].value_counts().sort_index())
```

```python
duplicate_rows = df[df.duplicated(keep=False)]

print("Rows involved in exact duplicates:")
display(duplicate_rows)

print("Number of rows involved:", len(duplicate_rows))
```

```python
from pathlib import Path

# Keep the original dataframe unchanged
df_clean = df.drop_duplicates().copy()

# Save the processed dataset
processed_dir = Path("../data/processed")
processed_dir.mkdir(parents=True, exist_ok=True)

output_path = processed_dir / "questionnaire_cleaned.csv"
df_clean.to_csv(output_path, index=False)

print("Original rows:", len(df))
print("Rows after removing exact duplicates:", len(df_clean))
print("Duplicates remaining:", df_clean.duplicated().sum())
print("Missing values:", df_clean.isnull().sum().sum())
print("Saved to:", output_path)
```

## Data Cleaning Decisions

- Loaded the original questionnaire dataset containing 465 rows and 16 columns.
- Standardized column headers by removing embedded newline and carriage-return characters.
- Identified two rows involved in one exact duplicate pair.
- Removed one duplicate copy, resulting in 464 rows and 16 columns.
- Confirmed that the cleaned dataset contains no exact duplicate rows and no missing values.
- Preserved the original CSV unchanged.
- Retained unusual age, height, blood-group coding, and period-duration values because their correct interpretation could not be confirmed.
- Did not convert response codes into Yes/No because the original coding scheme still needs verification.
- Saved the processed dataset as `data/processed/questionnaire_cleaned.csv`.
